## 1: Employee Performance Bonus Eligibility

```python
employees = {
            "Ravi": 92,
            "Anita": 88,
            "Kiran": 92,
            "Suresh": 85
            }

# This function calculates maximum score from the dictionary
max_score = max(employees.values())

# Empty list which will store the employe names having max score
eligible = []

# This for loops appends the employe names whose highest score is equal to max score
for key in employees:
    if employees[key] == max_score:
        eligible.append(key)
        
# Output variable contains final output String value
output = "Top Performers Eligible for Bonus:"

# String output is added with the employee names in eligible list.
for i in eligible:
    output += f"{i}, " 

# Again output string is added with another string with max score
output += f"(Score: {max_score})"

#Final output string is combination of every string. 
print(output)
```

## 2: Search Query Keyword Analysis

```python
query = "Buy mobile phone buy phone online"

# Convert query to lowercase
query = query.lower()

# Replaces any common punctuation with blank space 
query = query.replace(",", " ").replace("?"," ").replace(".", " ")

# Splits the query at " " spaces and returns a list keywords
keywords = query.split()

# Seperate the unique keywords using sets
unique_keywords = set(keywords)

# Empty dictionary which will contain final output (keyword : frequency)
key_freq = {}

# Nested for loop which counts the frequency of each unique keyword
for unique in unique_keywords:
    count = 0
    for words in keywords:
        if unique == words:
            count += 1
    # Add keywords in dictionary key_freq along with the frequency if more than 1
    if count > 1:
        key_freq[unique] = count
        
print(key_freq)            
```

## 3: Sensor Data Validation

```python
sensor_readings = [3, 4, 7, 8, 10, 12, 5]

# Emty dictionary which will contain final output
hour_value = {}

# Iterating index by index to find even readings and assigning it to dictionary
for i in range(len(sensor_readings)):
    if sensor_readings[i]%2 == 0:
        hour_value[i] = sensor_readings[i]

print(hour_value.items())
```

## 4: Email Domain Usage Analysis

```python
emails = [
        "ravi@gmail.com",
        "anita@yahoo.com",
        "kiran@gmail.com",
        "suresh@gmail.com",
        "meena@yahoo.com"
        ]
# Created an empty dictionary which will contain domain : count
domain_count = {}

# Adding elements to dictionary
for i in emails:
    domain = (i.split("@")[1])
    if domain in domain_count:
        domain_count[domain] += 1
    else:
        domain_count[domain] = 1
    
    
total_users = len(emails)

# Displaying and calculating percentage
for domain, count in domain_count.items():
    percentage = count/ total_users * 100
    print(f"{domain} : {int(round(percentage))}%")
```

## 5: Sales Spike Detection

```python
sales = [1200, 1500, 900, 2200, 1400, 3000]

# Compute Average of sales
average_sales = sum(sales)/len(sales)

# Base sales are 30% above average sales
base_sales = average_sales + 0.3*average_sales

# Iterate using enumerate function and check if the sales are above base_sales
for index, value in enumerate(sales):
    if value > base_sales:
        print(f"Day {index+1}: {value}")
```

## 6: Duplicate User ID Detection

```python
user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]

# Create a empty dictionary which will store user_id : counts pair
id_count = {}

# Adding elements to dictionary
for i in user_ids:
    if i in id_count:
        id_count[i] += 1
    else:
        id_count[i] = 1

# Check if counts are > 1 and display them
for ids, counts in id_count.items():
    if counts > 1:
        print(f"{ids} --> {counts} times")
```
