```python
import re       # regular expression
import string
import nltk

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# nltk does not come with the data preinstalled - we have to download it
nltk.download('stopwords')  # for filtering common word
nltk.download('wordnet')
```

#### Initialize tools - which we use them further

```python
# Set - for unique stopwords
stop_words = set(stopwords.words('english'))

# Object for lemmatizer
lemmatizer =  WordNetLemmatizer()
```

```python
text = 'I love this movie !!!'
print(f"Original Text : {text}")
```

```python
def preprocess_text(text):
    # 1. Lowercasing
    text = text.lower()
    
    # 2.Remove punctutaions (?, !, .)
    text = text.translate(str.maketrans('', '', string.punctuation))
    
    # 3. Remove Special characters
    text = re.sub(r'[^\w\s]', '', text)     # ^ : negation, \w : words, \s : space, [^\w\s] : anything expect words and spaces
    
    # 4. Tokenization
    tokens = text.split()
    
    # 5. Stopwords removal
    tokens = [word for word in tokens if word not in stop_words]    # list contaions words which are not in stopwords
    
    #  Lemmatization : convert to the root words --> (running = run)
    tokens = [lemmatizer.lemmatize(word) for word in tokens]
    
    return tokens
```

```python
processed = preprocess_text(text)
print(f'Preprocessed Text : {processed}')
```

```python
# To convert back tokens to sentance
def tokens_to_sentance(tokens):
    return ' '.join(tokens)
```

```python
example = preprocess_text('This course is amazing !!!')
print(example)
```
