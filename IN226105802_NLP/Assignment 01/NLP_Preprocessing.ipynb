```python
# Import Regular Expression Library and Counter
import re
from collections import Counter
```

#### Build Advanced preprocessing function

```python
def preprocess_text(text):
    
    # Step 1 : Lowercasing
    text = text.lower()
    
    # Step 2 : Removing Numbers
    text = re.sub(r'\d+', '', text)     # \d+ = "123", "45", "9" etc.
    
    # Step 4 : Removing repeated characters
    text = re.sub(r'(.)\1{2,}', r'\1\1', text)
    
    # Step 5 : Removing URL's & Emails
    text = re.sub(r'http\S+|www\S+', '', text)  # \S --> non-space character
    text = re.sub(r'\S+@\S+', '', text)
    
    # Removing Puncuations
    # [^\w\s] = anything that is NOT letter/number/space
    # \w --> letters + numbers  |   \s -->whtespaces    |   ^ --> NOT
    text = re.sub(r'[^\w\s]', '', text)
    
     # Step 3 : Removing Extra Spaces
    text = re.sub(r'\s+', " ", text).strip()    # \s --> spaces | strip() --> removes extra spaces of start and end
    
    # Tokenizing and Removing short words except 'not' and 'no'
    tokens = text.split()
    tokens = [words for words in tokens if len(words) > 2 or words in ['no', 'not']]
    text = " ".join(tokens) # Join the tokens again
    
    return text
```

```python
text = "I am sooo happy with this product!!!"

print(preprocess_text(text))
```

#### Test function with sample data

```python
sample_inputs = [ 
    "Get 100% FREE access now!!!", 
    "I absolutely looooved this product", 
    "Worst service ever... 0/10", 
    "Call me at 9876543210", 
    "This is THE best course!!!", 
    "Visit https://openai.com now!", 
    "Nooooo this is baaad!!!", 
    "OK OK OK I got it", 
    "Win $$$ now!!! Limited offer!!!", 
    "I am not happy with this" 
]

for inputs in sample_inputs:
    cleaned_text = preprocess_text(inputs)
    tokens = cleaned_text.split()
    print(f"Original : {inputs}\nTokens : {tokens}\nCleaned : {cleaned_text}\n\n")
```

#### Token Analytics

```python
for inputs in sample_inputs:
    cleaned_text = preprocess_text(inputs)
    tokens = cleaned_text.split()
    unique_tokens = len(set(tokens))    # set --> For unique
    total_tokens_length = sum(len(word) for word in tokens)
    avg_token_length = total_tokens_length / len(tokens) if tokens else 0
    
    print("-" * 50)
    print(f"Original : {inputs}")
    print(f"Tokens : {tokens}")
    print(f"Cleaned : {cleaned_text}")
    print(f"Total Tokens : {len(tokens)}")
    print(f"Unique Tokens : {unique_tokens}")
    print(f"Avg Token Length : {avg_token_length:.2f}")
```

```python
#  Analysis Questions Answers
# The sentence "OK OK OK I got it" had the most noise
# The sentence "I absolutely looooved this product" retained the most meaningful tokens
```

#### Frequency Analysis

```python
# Combine all tokens

all_tokens = []
for inputs in sample_inputs:
    cleaned_text = preprocess_text(inputs)
    tokens = cleaned_text.split()
    all_tokens.extend(tokens)

word_freq = Counter(all_tokens)

# Top 10 most frequent words
most_freq = word_freq.most_common(10)

# Top 5 least frequent words 
least_freq = sorted(word_freq.items(), key = lambda w : w[1])[:5]

print(f"Most Frequent : {most_freq}")
print(f"Least Frequent : {least_freq}")
```

#### Full Pipeline 

```python
def full_pipeline(text_list):
    
    # Handling Empty input list
    if not text_list:
        return {"tokens": [], "clean_sentences": []}
    
    all_tokens = []
    clean_sentences = []
    
    for text in text_list:
        cleaned_text = preprocess_text(text)
        clean_sentences.append(cleaned_text)
        
        # If cleaned_text is empty
        if not cleaned_text:
            continue
        
        tokens = cleaned_text.split()
        all_tokens.extend(tokens)
        
    return {"tokens":all_tokens, "clean_sentences":clean_sentences}

print(full_pipeline(sample_inputs))
```
