```python
import re
```

```python
def preprocessing_text(text):
    
    # Step 1 : Lowercasing
    text = text.lower()
    
    # Step 2 : Removing Numbers
    text = re.sub(r'\d+', '', text)     # \d+ = "123", "45", "9" etc.
    
    # Step 4 : Removing repeated characters
    text = re.sub(r'(.)\1{2,}', r'\1\1', text)
    
    # Step 5 : Removing URL's & Emails
    text = re.sub(r'http\S+|www\S+', '', text)  # \S --> non-space character
    text = re.sub(r'\S+@\S+', '', text)
    
    # Removing Puncuations
    # [^\w\s] = anything that is NOT letter/number/space
    # \w --> letters + numbers  |   \s -->whtespaces    |   ^ --> NOT
    text = re.sub(r'[^\w\s]', '', text)
    
     # Step 3 : Removing Extra Spaces
    text = re.sub(r'\s+', " ", text).strip()    # \s --> spaces | strip() --> removes extra spaces of start and end
    
    # Tokenizing and Removing short words except 'not' and 'no'
    tokens = text.split()
    tokens = [words for words in tokens if len(words) > 2 or words in ['no', 'not']]
    text = " ".join(tokens) # Join the tokens again
    
    return text
```

```python
text = "I am sooo happy with this product!!!"

print(preprocessing_text(text))
```
