This project demonstrates the implementation of a conversational chatbot using transformer-based models from Hugging Face. It explores two different models for different purposes:

- DialoGPT for conversational interaction
- FLAN-T5 for accurate question answering

The goal is to understand how transformer models work and how different models behave in different NLP tasks.

```python
# Install libraries

from transformers import AutoModelForCausalLM   # AutoModelForCausalLM : Generates Responses
from transformers import AutoTokenizer          # AutoTokenizer : Converts text --> numbers
import torch                                    # Pytorch : For backend
```

```python
# Make it use of GPU
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using Device:", device)
```

```python
# Load Model -  DialoGPT : Best for ChatBot
# DialoGPT is a conversational transformer model trained on large-scale dialogue data. It is designed to generate human-like responses and maintain conversational flow.

tokenizer = AutoTokenizer.from_pretrained("microsoft/DialoGPT-medium")      # Tokenizer Loads Vocabulary
model = AutoModelForCausalLM.from_pretrained("microsoft/DialoGPT-medium").to(device)   # Model Downloads from Hugging face
```

```python
# Load Model -  DialoGPT - Large : Upgraded for ChatBot 

tokenizer_large = AutoTokenizer.from_pretrained("microsoft/DialoGPT-large")      # Tokenizer Loads Vocabulary
model_large = AutoModelForCausalLM.from_pretrained("microsoft/DialoGPT-large").to(device)   # Model Downloads from Hugging face
```

```python
# Generate a response

# Input we have to give it to model
input_text = "Hello, how are you?"

# text --> token ids (numbers)
#  encodes( input_text and adds end_of_sentence token at the end of text so the model understands that sentence and ended) and return(pytorch tensors which model only understands)
input_ids = tokenizer_large.encode(input_text + tokenizer_large.eos_token, return_tensors='pt').to(device)

# Generates output with max 100 tokens of [input + output]
output = model_large.generate(input_ids, max_length=100)

# Creates response by Numbers --> Text  |   output[:, input_ids.shape[-1]:] = Skips input part, keeps oly generated response   |   skip_special_tokens = removes tags <eos>, <pad>...  |   [0] = extracts tensors from batch
response = tokenizer_large.decode(output[:, input_ids.shape[-1]:][0], skip_special_tokens=True)

# Print Response
print(response)     
```

#### Chat Loop & Conversation History

```python
chat_history_ids = None

while True:
    
    # Take user input
    user_input = input("You: ")
    
    # Handle condition for exit
    if user_input.lower() in ["exit", "quit"]:
        print("Chatbot: Goodbye !")
        break
    
    # Encode : Text --> Numbers = return pytorch tensors
    new_input_ids = tokenizer_large.encode(user_input + tokenizer_large.eos_token, return_tensors='pt').to(device)
    
    # Add chat history to input
    if chat_history_ids is not None:
    
        # Concatenate [History + user input]
        input_ids = torch.cat([chat_history_ids, new_input_ids], dim=-1)
    else:
        input_ids = new_input_ids

    # Create Attention Mask
    attention_mask = torch.ones(input_ids.shape, device=device)
    
    # Generate Output 
    chat_history_ids = model_large.generate(
        input_ids, 
        max_length = 1000, 
        pad_token_id = tokenizer_large.eos_token_id,
        do_sample = True,   # Enables Randomness
        temperature = 0.6,  # Controls Creativity
        top_k = 40,         # Model chooses top 50 probable words
        top_p = 0.85,         # Selects words until probability = 90%
        no_repeat_ngram_size = 3
        )
    
    # Limit History
    chat_history_ids = chat_history_ids[:, -512:]
    
    # Generate response : Decode = Numbers --> Text
    response = tokenizer_large.decode(chat_history_ids[:, input_ids.shape[-1]:][0], skip_special_tokens=True)
    
    # Print Chatbot responses
    print("Chatbot: ", response)
    
    
```

#### DialoGPT was used for conversational flow, but it lacks factual accuracy as it is trained on conversational data. To address this, another transformer model (FLAN-T5) is explored for more accurate and informative responses.

### FLAN-T5

FLAN-T5 is an instruction-tuned transformer model that performs well on tasks like question answering and explanation. It provides more accurate and structured responses compared to conversational models.

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
```

```python
flan_tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-large")
flan_model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-large").to(device)
```

```python
def ask_flan(question):

    prompt = "Answer clearly: "+ question

    inputs = flan_tokenizer(prompt, return_tensors='pt').to(device)

    outputs = flan_model.generate(
        **inputs,
        max_length = 100,
        do_sample = True,
        temperature = 0.7                         
        )

    response =  flan_tokenizer.decode(outputs[0], skip_special_tokens=True)

    return(response)
```

```python
while True:
    user_input = input("You: ")
    if user_input.lower() in ["exit", "quit"]:
        print("Flan: Goodbye !")
        break
    output = ask_flan(user_input)
    print("Flan: ",output)
```

### Observation

FLAN-T5 produces more accurate and informative responses. It performs better for factual and explanatory tasks but does not maintain conversational flow like DialoGPT.

## Conclusion

This project demonstrates how different transformer models can be used for different NLP tasks. DialoGPT is effective for building conversational chatbots, while FLAN-T5 is better suited for accurate question answering. Choosing the right model based on the task is crucial in AI system design.
