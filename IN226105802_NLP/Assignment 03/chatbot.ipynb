```python
# Install libraries

from transformers import AutoModelForCausalLM   # AutoModelForCausalLM : Generates Responses
from transformers import AutoTokenizer          # AutoTokenizer : Converts text --> numbers
import torch                                    # Pytorch : For backend
```

```python
# Make it use of GPU
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using Device:", device)
```

```python
print("cuda available:", torch.cuda.is_available())
print("torch version:", torch.__version__)
print("torch cuda version:", torch.version.cuda)
```

```python
# Load Model -  DialoGPT : Best for ChatBot

tokenizer = AutoTokenizer.from_pretrained("microsoft/DialoGPT-medium")      # Tokenizer Loads Vocabulary
model = AutoModelForCausalLM.from_pretrained("microsoft/DialoGPT-medium").to(device)   # Model Downloads from Hugging face
```

```python
# Generate a response

# Input we have to give it to model
input_text = "Hello, how are you?"

# text --> token ids (numbers)
#  encodes( input_text and adds end_of_sentence token at the end of text so the model understands that sentence and ended) and return(pytorch tensors which model only understands)
input_ids = tokenizer.encode(input_text + tokenizer.eos_token, return_tensors='pt').to(device)

# Generates output with max 100 tokens of [input + output]
output = model.generate(input_ids, max_length=100)

# Creates response by Numbers --> Text  |   output[:, input_ids.shape[-1]:] = Skips input part, keeps oly generated response   |   skip_special_tokens = removes tags <eos>, <pad>...  |   [0] = extracts tensors from batch
response = tokenizer.decode(output[:, input_ids.shape[-1]:][0], skip_special_tokens=True)

# Print Response
print(response)     
```

#### Chat Loop & Conversation History

```python
chat_history_ids = None

while True:
    
    # Take user input
    user_input = input("You: ")
    
    # Handle condition for exit
    if user_input.lower() in ["exit", "quit"]:
        print("Chatbot: Goodbye !")
        break
    
    # Encode : Text --> Numbers = return pytorch tensors
    new_input_ids = tokenizer.encode(user_input + tokenizer.eos_token, return_tensors='pt').to(device)
    
    # Add chat history to input
    if chat_history_ids is not None:
        # Concatenate [History + user input]
        input_ids = torch.cat([chat_history_ids, new_input_ids], dim=-1)
    else:
        input_ids = new_input_ids
    
    # Generate Output 
    chat_history_ids = model.generate(input_ids, max_length = 1000, pad_token_id = tokenizer.eos_token_id)
    
    # Generate response : Decode = Numbers --> Text
    response1 = tokenizer.decode(chat_history_ids[:, input_ids.shape[-1]:][0], skip_special_tokens=True)
    
    # Print Chatbot responses
    print("Chatbot: ", response1)
    
    
```
