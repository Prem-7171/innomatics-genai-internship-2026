## 1: Social Media – Post Engagement Analyzer

```python
# Likes from different region from different timelines
likes = [200, 150, 300, 400]

# Calculate total likes
def calc_total_likes(likes):
    likes_total = 0
    for like in likes:
        likes_total += like
    return likes_total


# Function to know status
def analyze_post(total_likes):
    if total_likes >= 1000:
        return "Viral Post"
    else:
        return "Normal Engagement"
        
total_likes = calc_total_likes(likes)
status = analyze_post(total_likes)

print(f"Likes: {total_likes}, Status: {status}")
```

## 2: Healthcare – Medicine Stock Alert System

```python
inventory = {
    "Paracetamol": 6,
    "Aspirin": 15,
    "Insulin": 4
}

def moniter_stock(inventory):
    '''Returns a dictionary having status report'''
    status_report = {}
    
    for medicine, stock in inventory.items():

        # If stocks are low, Assign the medical with Stock alert message
        if stock < 10:
            status_report[medicine] = "Status: Low stock Alert!"
        else:
            status_report[medicine] = "Stock Sufficient"

    return status_report

def main():
    # store the status_report dictionary into report
    report = moniter_stock(inventory)
    # Output using loop
    for medicine, status in report.items():
        print(f"{medicine} | stock: {stock}")

main()
```

## 3: Agriculture – Rainfall Adequacy Checker

```python
# List contaioning railfall values
rainfall = [60, 75, 80, 70, 65]

# Required level of rainfall
required = 70


def rainfall_checker(rainfall, required):
    """Returns Average rainfall and rainfall Status"""

    # Computing average
    average_rainfall = sum(rainfall)/len(rainfall)

    
    if average_rainfall >= required:
        status = "Adequate Rainfall"
        return status, average_rainfall
    else:
        status = "Inadequate Rainfall"
        return status, average_rainfall

def main():
    status, avg = rainfall_checker(rainfall, required)
    print(f"Average Rainfall: {avg}")
    print(f"Rainfall Status: {status}")

main()
        
```

## 6: Agriculture – Premium Crop Price Filter

```python
# Created a function to filter the crop prices
def price_filter(crop_prices, threshold):
    """Returns list of premium crops"""

    # Empty list which will store the premium crop prices
    premium_crops = []

    # iterate through every crop price and append to list which are > 2000
    for price in crop_prices:
        if price > threshold:
            premium_crops.append(price)

    return premium_crops

def main():

    # Sample crop prices data in list
    crop = [1500, 2500, 2100, 1600, 5000, 3150]
    threshold = 2000
    
    Premium_crops = price_filter(crop, threshold)

    print("Premium crops:",Premium_crops)

main()
```

## 9: E-Learning – Student Attendance Eligibility System

```python
attendence = ['A','A','P','A','P','A','P','P']


def attendence_checker(records):
    """Returns if Eligible or Not"""
    count = 0
    # For loop to count number of "P" (Count stores no. of p)
    for remarks in records:
        if remarks == "P":
            count += 1

    # Calculate Percentage 
    percentage = int(round(count/len(records) * 100))

    # Eligibility Criteria
    if percentage >= 75:
        return "Eligible", percentage 
    else:
        return "Not Eligible" ,percentage

stat , percent = attendence_checker(attendence)
print(f"Attendence Percentage :{percent}%")
print(f"Exam Eligibility: {stat}")
```

## 10: Smart Electricity Bill Analyzer

```python
def bill_calculater(units):
    bill = 0

    if units <= 100:
        bill = units*3
    elif units > 100 and units < 200:
        bill = 300 + (units - 100) * 5
    else:
        bill = 800 + (units - 200) * 7

    if bill < 500:
        status = "Low usage"
    elif bill >= 500 and bill < 1500:
        status = "Moderate usage"
    else:
        status = "High usage"

    return bill, status
    
def main():
    units = 202
    bill, status = bill_calculater(units)
    print(f"Bill for {units} units : {bill}")
    print(f"Usage Status: {status}")

main()
```
