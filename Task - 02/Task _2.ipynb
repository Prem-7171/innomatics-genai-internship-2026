##  1: Unique Words in a Sentence

```python
sen = input("Enter a sentance: ")

# split() function splits the string with specific operator and returns the list.
words = sen.split()

# words is a list of words in the sen, it gets converted into set. As set data structure ignores duplicate values.
set_words = set(words)

print("Unique words count: ", len(set_words))
print("Unique words: ",set_words)
```

## 2: Highest Salary from Employee Data

```python
employees = {
    "Ravi": 75000,
    "Anita": 68000,
    "Kiran": 72000
}

# to find key having maximum value in the dictionary - max(dictionary, key = dictionary.get)
highest = max(employees, key = employees.get)

print(f"Highest Salary: {highest} - {employees[highest]}")
```

## 3: Find Maximum and Minimum Values

```python
numbers = [45, 22, 89, 10, 66, 95, 21, 34]

# First we initialize the highest and lowest as the 1st number of the list
highest = numbers[0]
lowest = numbers[0]

# iterate through each number from the list
for i in numbers:

    # if i >= highest assign 'i' in the highest. this continues till the end of list
    if i >= highest:
        highest = i

    # if i <= lowest assign 'i' in the lowest. this continues till the end of list
    if i <= lowest:
        lowest = i

# At the end of the loop, Highest and Lowest contains the max and min numbers from the list
print("List: ", numbers)
print("Max: ", highest)
print("Min: ", lowest)
```

## 4: Count Products Above a Price Threshold

```python
prices = [450, 1200, 899, 1500, 300]

# Variable having Price Threshold Value
price_thresh = 1000

# Loop through each element and increment count for prices > price_threshold
count = 0
for price in prices:
    if price > price_thresh:
        count += 1

print(f"Products above {price_thresh}: {count}")
```

## 5: Calculate Attendance Percentage

```python
attendance = ["P", "P", "A", "P", "P"]

# Count "P" in attendance list
count = 0
for i in attendance:
    if i == "P":
        count += 1

# Percentage Formula
percentage = count/len(attendance) * 100

print("Attendance Percentage:", percentage)
```
