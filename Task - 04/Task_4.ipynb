## 1: Mobile Recharge Validation System

```python
plans = (199, 299, 399, 599)

# Define a function that checks if the recharge is valid.
def validate_recharge(amount, plans):
    """Returns True or false if amount is valid or not"""
    if amount in plans and >= 50:
        return True
    else:
        return False


# Loops till the recharge amount entered is valid and prints if recharge is successful.
while True:
    amount = int(input("Enter Recharge Amount: "))
    if validate_recharge(amount, plans):
        print(f"Recharge of {amount} /- successful !")
        break
    else:
        print("Please enter Valid plan. Try again.")
        
```

## 2: Inventory Reorder Alert System

```python
inventory = {
    "Rice": 20,
    "Wheat": 10,
    "Sugar": 5
}

# Sets a threshold value for stock alert
threshold = 15

# A function which takes threshold and inventory dictionary as parameters to check the stocks.
def check_inventory(inventory, threshold):
    for product, stock in inventory.items():
        if stock < threshold:
            print(f"Reorder Alert for {product}")
        else:
            print(f"Stock OK for {product}")

# Calling Function
check_inventory(inventory, threshold)
```

## 3: Student Result Processing System

```python
marks = [15, 45, 95, 86, 73, 25, 66]


# Function calculates average and checks if fail or pass
def result(marks):
    """Returns average and Result from the passed list"""
    total = 0
    result = ""
    
    for m in marks:
        total += m

    average = round(total/len(marks), 2)

    if average >= 50:
        result = "Pass"
    else:
        result = "Fail"

    return average, result

# The return is in form of tuple, So stored in two variables
avg, status = result(marks)
print(f"Average: {avg}, Result: {status}")
```

## 4: Cab Fare Estimator with Retry Option

```python

def fare(distance, peak_hour):
    """Returns fare"""
    #  Assign variables for base fare and per km charges as they can be changed in future
    base_fare = 50
    charge = 12

    # Formula for calculating fare
    fare = base_fare + charge*distance
    
    if peak_hour == "yes":
        # for peak hour 25% more fare
        fare = fare + fare*0.25
        return fare
    else:
        return fare

def main():
    distance = int(input("Input Total Distance covered in a trip: "))
    peak_hour = input("Peak Hour (yes/no)").lower()
    print(f"Your total fare is : {fare(distance, peak_hour)}")

while True:
    main()
    retry = input("DO you want to retry? (yes/no) :")
    if retry == "no":
        break
```

## 5: Employee Attendance Eligibility Checker

```python
attendence = ['A','A','P','A','P','A','P','P','A','P','A','P','A','A','P']


def attendence_checker(records):
    """Returns if Eligible or Not"""
    count = 0
    # For loop to count number of "P" (Count stores no. of p)
    for remarks in records:
        if remarks == "P":
            count += 1

    # Calculate Percentage 
    percentage = int(round(count/len(records) * 100))

    # Eligibility Criteria
    if percentage >= 75:
        return "Eligible", percentage 
    else:
        return "Not Eligible" ,percentage

stat , percent = attendence_checker(attendence)
print(f"{percent}% : {stat}")
```

## 6: Password Strength Checker

```python
password = "abc123"


def pass_strength_check(password):
    '''Returns True if password is strong , else returns error list'''

    # Empty list which stores errors
    errors = []

    # Initialize digits and speacial character as false at the start
    is_digit = False
    is_spl = False

    # Loop through each character from the password and assign true if at least 1 digit or speacial character is found
    for ch in password:
        
        if ch.isdigit():
            is_digit = True
            
        if ch in "@#$":
            is_spl = True

    # Append error to error list if Length <8
    if len(password) < 8:
        errors.append("Length must be greater than 8.")

    # Append error to error list of no digit found
    if is_digit == False:
        errors.append("There must be atleast 1 digit.")

     # Append error to error list of no special character found
    if is_spl == False:
        errors.append("Length must be atleast 1 speacial character.")

    # check if error list is empty or not, if not return error list
    if not errors:
        return True, errors
    else:
        return False, errors


strong, error = pass_strength_check(password)

if strong == False:
    for i in error:
        print(i)
else:
    print("Your Password is strong")
```
